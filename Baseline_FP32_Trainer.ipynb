# **YOLO Drone Detector Implementation**

Load the model and train with augmentations

```python
from ultralytics import YOLO

model = YOLO('yolo11n.pt')

model.train(
    data = "/content/YOLO_Drone_Detector-1/data.yaml",
    epochs = 100,
    imgsz = 640,
    batch = 32,
    workers = 16,
    save = True,
    cache = 'disk',
    plots = True,
    task='detect',

    # Data Augmentation for better model generalization
    mosaic=0.7,
    mixup=0.05,
    degrees=15,
    translate=0.1,
    scale=0.7,
    fliplr=0.5,
    patience=10,
    cos_lr=True,
    amp=True,

    # Use GPU for training
    device='cuda'
)
```

Metrics, Results and Export

```python
# Validate the model
metrics = model.val()

print(f"Mean Average Precision @ 50-95: {metrics.box.map:.3f}")
print(f"Mean Average Precision @ 50: {metrics.box.map50:.3f}")
```

Further testing and validaion

```python
model = YOLO("/content/runs/detect/train/weights/best.pt")

# Validation
metrics = model.val()

print(metrics.box.map)
print(metrics.box.map50)
print(metrics.box.map75)
```
